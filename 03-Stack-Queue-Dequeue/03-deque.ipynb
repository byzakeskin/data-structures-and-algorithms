```python
class Deque:
    def __init__(self):
        self.items = []

    def addRight(self, item):
        self.items.append(item)

    def addLeft(self, item):
        self.items.insert(0, item)

    def removeRight(self):
        return self.items.pop()

    def removeLeft(self):
        return self.items.pop(0)

    def isEmpty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)
```

```python
myDeque = Deque()
```

```python
myDeque.addRight(1)
```

```python
myDeque.addRight(2)
```

```python
myDeque.addLeft(3)
```

```python
myDeque.addLeft(4)
#[4, 3, 1, 2]
```

```python
myDeque.addRight(5)
#[4, 3, 1, 2, 5]
```

```python
myDeque.size()  # 5
```
