Queue: Ordered, FIFO, Front & Rear

```python
'''
Queue()

enqueue(item)
dequeue()
size()
isEmpty()
'''
```

```python
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.insert(0,item)

    def dequeue(self):
        return self.items.pop(0)

    def size(self):
        return len(self.items)

    def isEmpty(self):
        return len(self.items) == 0
```

```python
myQueue = Queue()
```

```python
myQueue.isEmpty()  # True
```

```python
myQueue.enqueue(1)
myQueue.enqueue(2)
myQueue.enqueue(3)
```

```python
myQueue.size()  # 3
```

```python
myQueue.dequeue()  # 3
```
