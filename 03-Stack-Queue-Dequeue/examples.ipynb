Implement Stack Using Queue

```python
def __init__(self):
        self.myQueue = deque()

def push(self, x: int) -> None:
        self.myQueue.append(x)

def pop(self) -> int:
        for i in range(len(self.myQueue) - 1):
            self.myQueue.append(self.myQueue.popleft())
        return self.myQueue.popleft()

def top(self) -> int:
    return self.myQueue[-1]

def empty(self) -> bool:
    return len(self.myQueue) == 0                        
```

Baseball Game

```python
myInput = ["5", "2", "C", "D", "+"]
```

```python
def solution():

    myStack = []

    for op in myInput:
        if op == "C":
            myStack.pop()
        elif op == "D":
            myStack.append(myStack[-1] * 2)
        elif op == "+":
            myStack.append(myStack[-1] + myStack[-2])
        else:
            myStack.append(int(op))
    return sum(myStack)
```

```python
solution()
```

Daily Temperatures

```python
temperatures = [73, 74, 75, 71, 69, 72, 76, 73]
```

```python
def solution():
    result = [0] * len(temperatures)
    myStack = []

    for ix, temprature in enumerate(temperatures):
        while myStack and temprature > myStack[-1][0]:
            stackTemp, stackIndex = myStack.pop()
            result[stackIndex] = ix - stackIndex
        myStack.append([temprature, ix])
    return result
```

```python
solution()
```
