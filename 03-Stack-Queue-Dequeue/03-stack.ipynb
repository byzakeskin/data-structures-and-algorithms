Stack: Ordered, LIFO, Top & Base

```python
'''
Stack()

push(item)
pop()
showLast()
size()
isEmpty()

'''
```

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def showLast(self):
        return self.items[-1]

    def size(self):
        return len(self.items)

    def isEmpty(self):
        return len(self.items) == 0
```

```python
myStack = Stack()
```

```python
myStack.isEmpty()  # True
```

```python
myStack.push(1)
```

```python
myStack.push(2)
```

```python
myStack.push(3)
```

```python
myStack.showLast()  # 3
```

```python
myStack.pop()  # 3
```
