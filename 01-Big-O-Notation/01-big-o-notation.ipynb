```python
#Big O Notation -> O()
#Time Complexity - Space Complexity 
#https://www.bigocheatsheet.com/
```

```python
def bigon (n):
    for i in range(0,n):
        print(i)
```

```python
bigon(5)
```

```python
def bigon2 (n):
    for i in range(0,n):
        for j in range(0,n):
            print(i,j)
```

```python
bigon2(5) # O(n^2) - Quadratic Time Complexity
```

```python
import math
def log(n):
    while n >1:
        n = math.floor(n/2)
        print(n)
```

```python
log(100) # O(log n) - Logarithmic Time Complexity
```

```python
def nlogn(n):
    lim = n
    while n >1:
        n = math.floor(n/2)
        for i in range(0,lim):
            print(i)
```

```python
nlogn(4) # O(n log n) - Log Linear Time Complexity
```

```python
def nfactorial (n):
    if n == 0:
        print("1")
        return
    else:
        for i in range(0,n) :
            print(i)
            nfactorial(n-1) # recursive call
```

```python
nfactorial(3) # O(n!) - Factorial Time Complexity
```

```python
counter = 0
def actualFactorial(n):
    global counter
    counter += 1
    print("Counter: ", counter)
    if (n == 1 or n == 0):
        return 1
    else:
        return (n * actualFactorial(n - 1))
```

```python
actualFactorial(4)
```
