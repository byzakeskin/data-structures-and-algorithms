```python
#Contains Duplicate
#Given an integer array nums, return true if any value appears at least twice in the array, and return false if every element is distinct.
myList = [1,2,3,4,5,6,7,8,1]
```

```python
#Time Complexity: O(n), Space Complexity: O(n)
def solution():
    hashSet = set()
    for num in myList:
        if num in hashSet:
            return True
        hashSet.add(num)
    return False
```

```python
solution()
```

```python
def solution2():
    return len(myList) != len(set(myList))
```

```python
solution2()
```

Single Number

Given an non-empty array of integers nums, every element appears twice except for one. Find that single one.

Edge Cases:

If the array has only one element, return that element.

if len(nums) == 1:
    return nums[0]

```python
def singleNumber():
    myList = [3,5,7,4,3,7,5,4,1]
    result = 0
    for num in myList:
        result ^= num
    return result
```

```python
singleNumber()
```

Majority Element

Given an array nums of size n, return the majority element. The majority element is the element that appears more than ⌊n / 2⌋ times. 
You may assume that the majority element always exists in the array.

```python
myList = [2,3,4,1,1,1,1,1,5]
```

```python
def findMajorityElement():
    numbers = {}
    result = 0
    maxNumber = 0

    for num in myList:
        numbers[num] = 1 + numbers.get(num, 0)
        if numbers[num] > maxNumber:
            result = num
        maxNumber = max(maxNumber, numbers[num])
        return result    
```

```python
findMajorityElement()
```

Boyer Moore Algorithm

```python
def boyerMoore():
    result = 0
    count = 0

    for num in myList:
        if count == 0:
            result = num
        count += (1 if num == result else -1)
    return result
```
